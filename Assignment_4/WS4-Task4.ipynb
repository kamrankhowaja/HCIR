#### With the help of a diagram illustrate and explain the processes and pathways in a "conceptually complete" multimodal architecture for socially interactive agents.



#### 1. The Three Core Pillars (Columns)

The architecture divides the lifecycle of social interaction into three distinct operational columns:

Processing (Understanding Input): This pillar transforms raw user data into social understanding through a bottom-up hierarchy:

- Sensing ("receive"): The low-level capture of multimodal inputs (e.g., audio, video, text).

- Recognition ("find"): Extracting specific social features or patterns, such as identifying a facial expression, a gesture, or specific keywords.

- Interpretation ("understand"): Synthesizing the recognized features to infer the deeper social meaning, intention, or emotional state of the user.

Mapping (Action Selection): This pillar sits at the center of the architecture, determining how the agent should respond based on varying cognitive depths:

- Reactive (rules): Immediate, instinctual mappings that bypass deep thought (e.g., quickly turning toward a sudden sound).

- Associative (selection): Pattern-matching and learned behaviors that link recognized situations to appropriate social responses.

- Deliberative (planning): High-level cognitive processing that plans complex, long-term interaction strategies aligned with the agent’s goals and social norms.

Generation (Producing Output): This pillar translates the chosen response into physical expression through a top-down hierarchy:

- Formulation ("compose"): Planning the abstract content of the response (e.g., deciding the words to say and the general accompanying gestures).

- Realization ("create"): Generating the specific physical trajectories or motor commands (e.g., animating the avatar or synthesizing the text-to-speech audio).

- Execution ("perform"): The actual real-time delivery of the multimodal behavior back to the user.

#### 2. Hierarchical Processing Levels

The vertical structure of the figure reveals a clear distinction between the complexity of behaviors:

Lower Level (Sensing & Motor Behavior): Handled at the bottom of the diagram (Sensing $\leftrightarrow$ Reactive $\leftrightarrow$ Execution). This layer manages fast, real-time physical coordination, basic reflexes, and direct sensorimotor tracking.

Higher Level (Conversational & Socio-Relational Functions): Handled at the top of the diagram (Interpretation $\leftrightarrow$ Deliberative $\leftrightarrow$ Formulation). This layer governs the long-term context of the conversation, relationship building, empathy, and adherence to complex social etiquette.

#### 3. Multiple Routes & Information Pathways

The arrows connecting the blocks in the figure demonstrate that a socially interactive agent does not just process information in a strict, linear order. Instead, it utilizes multiple pathways concurrently:

The Quick Reactive Route (Dual-Route Architecture): Information can flow directly from Sensing $\rightarrow$ Reactive rules $\rightarrow$ Execution. This allows the agent to give instantaneous feedback (like nodding while someone is talking or blinking) without waiting for slow, high-level deliberation.

The Deep Deliberative Route: Information travels all the way up to Interpretation, over to Deliberative planning, and down through Formulation for complex reasoning.

Multidirectional & Incremental Flow: The bidirectional vertical arrows indicate that higher levels can guide lower levels (top-down expectations), while lower levels continuously update higher levels (bottom-up data). This enables incremental processing, meaning the agent can start formulating or adjusting its response mid-sentence as it receives new information from the user in real-time.
