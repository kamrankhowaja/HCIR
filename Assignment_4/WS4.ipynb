#### Name at least two characteristics of each of the three algorithms: TAMER, SABL and COACH. (10%)

TAMER:
- Greedy Learning - it assumes when a user gives feedback , it has already factored it in the long term consequneces of action , so it only seeks maximum imediate effect rather than a discounted sum of future rewards.
- Supervised Learning - a learning function that allows how much to learn from feedback, where the state action pairs serves as the inputs and human scalar feedback serves as the labels.

SABL:
- Trainer Strategy Modeling - uses diferent strategies and combination (eg Positive reward , positive punishment , negative reward , negative punishment) that determins the feedback.
- Informative absense of Feedback - it understads the trainer strategy , can create lack of feedback as impict signal, like in case of reward focused trainer, a silent pause is statistically interpreted by the agent as impicit negative feedback.

COACH:
- Policy Dependent advanatge function - COACH is built on the psychological insights that human feedback changes based on what agent already knows, so it takes human feedback as an advantage function , measuring how better and worse could be the action related to agents current policy.
- Direct Policy Gradient updates - As human is a critic , COACH uses the feedback to make direct updates to the agents parameterized policy using eligibility traces.
