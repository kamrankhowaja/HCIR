### Designing a Human-Robot Interaction Experiment (45%)

Suppose that you are developing a scenario where the Atlas robot should assist a plumber by bringing necessary tools and objects when requested. Let’s say, you have developed several ways in which the robot can handover an object to the plumber. Now, you would like to evaluate these by conducting an empirical study with experienced plumbers. (You can watch this video for inspiration: https://www.youtube.com/watch?v=-e1_QhJ1EhQ)

*Please refer to the following paper to understand the characteristics of good research questions and hypotheses*

**G. Hoffman and X. Zhao, “A Primer for Conducting Experiments in Human–Robot Interaction,” ACM Transactions on Human-Robot Interaction, vol. 10, no. 1, Art. no. 6, pp. 1–31, Oct. 2020, doi: 10.1145/3412374.**

#### Define at least one research question that you would like to investigate in the context of human-robot interaction in the above scenario.

How does the robot’s handover strategy (basic, communicative, adaptive) affect task performance, perceived safety, and trust of experienced plumbers during a tool-delivery task with the Atlas robot?

#### Identify the constructs involved in this research question(s).

Robot Behavior (Independent Construct)
- Handover strategy
- Basic reactive handover
- Communicative handover
- Adaptive handover

Human Experience (Dependent Constructs)
- Task performance
- Perceived safety
- Trust in robot
- Cognitive workload
- User preference / satisfaction

#### Formulate at least one hypothesis based on the research question(s) and the identified constructs.

H1 (Performance)
- Plumbers will complete tool handovers faster with the adaptive handover strategy compared to the basic reactive strategy.

H2 (Safety)
- Perceived safety will be higher in the communicative and adaptive conditions than in the basic condition.

H3 (Trust)
- Trust in the robot will be highest in the adaptive handover condition.

#### How would you operationalise the constructs in the above hypotheses?

1. Task Performance
    - Handover time (seconds)
    - Total task completion time
    - Number of failed handovers or drops
2. Perceived Safety
    - Questionnaire rating (e.g., 1–7 Likert scale)
    - Example item: “I felt safe interacting with the robot”
3. Trust
    - Standard HRI trust scale (Likert-based)
    - Example item: “I trusted the robot to deliver tools correctly”
4. Cognitive Workload
    - NASA-TLX questionnaire
    - Measures mental demand, effort, frustration
5. User Preference
    - Ranking of conditions after experiment
    - Post-study interview

#### How would you design your study (context, conditions, allocation of subjects)? Justify your study design.

Context
- Simulated plumbing workshop environment
- Realistic sink/pipe setup
- Atlas robot acts as assistant delivering toolsConditions (Independent Variable)

Within-subject comparison of 3 handover strategies:

Basic reactive
- Robot simply gives object when asked
Communicative
- Robot speaks + gives cues + announces tool
Adaptive
- Robot adjusts posture, timing, and orientation to user

Participants
- 20–30 experienced plumbers
- Professional background required

Allocation of Subjects
- Design type:
    - Each participant experiences all 3 conditions.

- Why this design?
    - Plumbers are a specialized population so hard to recruit many
    - Reduces variability between individuals
    - Improves statistical power
    - Allows direct comparison of conditions per person

#### What would your null hypothesis look like, for each of the hypotheses that you formulated above?

H0₁ (Performance)
- There is no significant difference in task completion time between basic, communicative, and adaptive handover strategies.

H0₂ (Safety)
- There is no significant difference in perceived safety between the different handover strategies.

H0₃ (Trust)
- There is no significant difference in trust ratings across the three handover strategies.
